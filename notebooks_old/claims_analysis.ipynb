```python
import json
import pandas as pd
import os
```

```python
file_path = "/home/adeluci2/projects/multidoc_summarization/results/ground_truth/multinews_plus/val/core_output.jsonl.demo"
claims = []
with open(file_path, "r") as f:
    for line in f:
        claims.append(json.loads(line))
```

```python
for c in claims:
    for c2 in c.get("meta", {}).get("claims", []):
        print(c2)
```

```python
for i, c in enumerate(claims):
    if c.get("score", -1) == 0:
        print(f"""{c['id']} {c['score']:.2}
        {c['decoded']}
        {c['meta']}
-----------------
""")
    if i > 10:
        break
```
