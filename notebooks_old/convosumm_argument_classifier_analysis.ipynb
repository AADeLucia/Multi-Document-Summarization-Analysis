```python
import os

import pandas as pd
import jsonlines

pd.set_option('display.precision', 4)
```

```python
stats = []
path = "/home/adeluci2/projects/multidoc_summarization/models/ConvoSumm-ConvoSumm/bin"
for dataset in ["multinews", "convosumm/reddit", "multixscience"]:
    with jsonlines.open(os.path.join(path, dataset, "test.jsonl")) as reader:
        for item in reader.iter():
            stats.append({
                "dataset": dataset,
                "id": item["id"],
                "sentences": len(item["texts"]),
                "arguments": sum([1 for i in item["predictions"] if i!=2])
            })

df = pd.DataFrame(stats)
df["claims"] = df.arguments / df.sentences
df["removed"] = 1 - df.claims
```

```python
df.groupby("dataset").agg({
    "claims": ["mean", "std"],
    "removed": ["mean", "std"]
})
```
